#IMPORTAMOS CONJUNTO DE DATOS DIRECTO DE KAGLE
[Conjunto de datos de redes sociales](https://www.kaggle.com/datasets/jocelyndumlao/social-media-and-consumer-behavior-2025)

```python

## Descargar dataset
path = kagglehub.dataset_download("jocelyndumlao/social-media-and-consumer-behavior-2025")
print("Path to dataset files:", path)

# Listar el contenido inicial
contenido_inicial = os.listdir(path)
print("\nContenido inicial (debería ser la subcarpeta):", contenido_inicial)

# Detectar subcarpeta automáticamente
if contenido_inicial:
    nombre_subcarpeta = contenido_inicial[0]
    nueva_ruta = os.path.join(path, nombre_subcarpeta)
    print(f"\nSe ha detectado una subcarpeta: {nombre_subcarpeta}")
else:
    print("\nERROR: no se encontró contenido en la ruta descargada.")
    raise SystemExit

# Listar el contenido dentro de la subcarpeta
archivos_finales = os.listdir(nueva_ruta)
print("\nArchivos dentro de la subcarpeta:")
for archivo in archivos_finales:
    print(" -", repr(archivo))   # repr() muestra caracteres especiales

# Buscar automáticamente el CSV (con o sin acentos)
csv_encontrado = None
for archivo in archivos_finales:
    if archivo.lower().endswith(".csv"):
        csv_encontrado = archivo
        break

if not csv_encontrado:
    print("\nERROR: No se encontró ningún archivo CSV en la subcarpeta.")
    raise SystemExit

print(f"\nCSV detectado: {repr(csv_encontrado)}")

# Construir ruta al CSV real
ruta_completa_csv = os.path.join(nueva_ruta, csv_encontrado)

# Cargar CSV en DataFrame
try:
    df_redesSociales = pd.read_csv(ruta_completa_csv)
    print("\n¡Carga exitosa del DataFrame!")
except Exception as e:
    print("\nERROR al cargar el CSV:")
    print(e)
```
